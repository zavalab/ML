```python
import numpy as np
import networkx as nx
import os
import imageio.v3 as iio
import tempfile
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib import colors

import torch
import random
from pyomo.environ import SolverFactory, value

import sys
import os
parent_dir = os.path.abspath(os.path.join(os.path.dirname('ProCatPy'), '..'))
sys.path.insert(0, parent_dir)
from ProCatPy.ProCatSystemFixedFreq import ProCatModel
#from ProCatPy.ProCatOpt import ProCatBO

import warnings
warnings.filterwarnings("ignore")
```

```python
# Simulate reaction network
seed = 7
np.random.seed(seed)
torch.manual_seed(seed)
random.seed(seed)

ebounds = np.array([0.8,1.6])
BEParams = np.array([[2.0],
                     [1.4]])
EaParams = np.array([[0.6],
                     [1.0]])
ns = 1

no = 101
Offsets = np.linspace(ebounds[0],ebounds[1],no).reshape(-1)
ystatic = np.zeros([no])

kfs = np.zeros([no,3])
krs = np.zeros([no,3])

thetaA = np.zeros(no)
thetaB = np.zeros(no)

dcn = ProCatModel(BEParams,EaParams)
dcn.SetEnergyParams(ebounds,ns,'Square')

for i in range(no):
    #print(i)
    eparams = {'Amplitude':[0.0],
               'DutyCycle':[0.0],
               'Frequency':1e-1,
               'Offset': Offsets[i],
               }
    dcn.SetSignalParams(eparams,False)
    dcn.solveBVP(1,1,'F','Thermo')
    ystatic[i] = dcn.calc_TOF(dcn.CBg)
    pt = dcn.timeParams(0)
    kfs[i,:] = pt.kf
    krs[i,:] = pt.kr
    thetaA[i] = dcn.CA[0]
    thetaB[i] = dcn.CB[0]

sindex = np.argmax(ystatic)
sX = Offsets[sindex]
sy = ystatic[sindex]
```

```python
fig, ax = plt.subplots(figsize = (8,6))
ax.plot(Offsets,thetaA)
ax.plot(Offsets,thetaB)
ax.legend(['A*','B*'])
ax.grid()
fig.tight_layout()
ax.tick_params(axis='both', which='major', labelsize=14)
fig.savefig('ABSurfConc.png',transparent=True)
```

```python
# Print TOF graph with stages shown
emin = 0.80
emax = 1.60
p1 = np.polyfit(Offsets[:25],np.log10(ystatic[:25]),1)
TOF_rhs = 10**(p1[0]*(emax) + p1[1])

p2 = np.polyfit(Offsets[-25:],np.log10(ystatic[-25:]),1)
TOF_lhs = 10**(p2[0]*(emin) + p2[1])

tofm = np.max([dcn.calc_TOF(dcn.CBg),TOF_rhs,TOF_lhs])
tofmin = 10**(np.floor((np.min(np.log10(ystatic))-1e-4) / 1) * 1)
tofmax = 10**(np.ceil((np.log10(tofm)+1e-4) / 1) * 1)
elhs = (np.log10(tofmax)-p1[1])/p1[0]
erhs = (np.log10(tofmax)-p2[1])/p2[0]

fig, ax = plt.subplots()
ax.semilogy(Offsets,ystatic,'b')
ax.semilogy([Offsets[0],elhs],[ystatic[0],tofmax],'r--')
ax.semilogy([Offsets[-1],erhs],[ystatic[-1],tofmax],'r--')
ax.semilogy([emin,emax],[TOF_lhs,TOF_rhs],'k',marker='*')
ax.set_xlim([ebounds[0]*0.99,ebounds[-1]*1.01])
ax.set_ylim([tofmin,tofmax])
ax.grid()
fig.tight_layout()
```

```python
eparams = {'Amplitude':[1.0],
           'DutyCycle':[0.15],
           'Frequency':1e4,
           'Offset': 1.1,
           }
dcn.SetSignalParams(eparams,True)
dcn.solveBVP(100,1,'F','Thermo')
# dcn.printSimResults()
# dcn.print_TOF(dcn.CBg)
# print(dcn.calc_TOF(dcn.CBg))
```

```python
# Get energy data
tplot = dcn.tdata
_, eplot = dcn.EnergyProgram(dcn.tdata)

emin = np.floor((np.min(eplot)-1e-4) / 0.1) * 0.1
emax = np.ceil((np.max(eplot)+1e-4) / 0.1) * 0.1

Offset = eparams['Offset']
idlist = np.argsort(eparams['DutyCycle'])
DCs = np.array(eparams['DutyCycle'])[idlist]
Amps = np.array(eparams['Amplitude'])[idlist]
timelist = np.concatenate([[0],np.sort(eparams['DutyCycle']),[1]])
nstage = len(timelist)
Elist = np.zeros(nstage-1)

for i in range(nstage-1):
    Elist[i] = Offset + np.sum(Amps[:i])/2 - np.sum(Amps[i:])/2
Elist = np.clip(Elist,ebounds[0],ebounds[1])
bsln = np.clip(Offset - np.sum(Amps)/2,ebounds[0],ebounds[1])

# Get static data and TOF graph info
tofmin = 10**(np.floor((np.min(np.log10(ystatic))-1e-4) / 0.1) * 0.1)

p1 = np.polyfit(Offsets[:25],np.log10(ystatic[:25]),1)
TOF_rhs = 10**(p1[0]*(np.max(eplot)) + p1[1])

p2 = np.polyfit(Offsets[-25:],np.log10(ystatic[-25:]),1)
TOF_lhs = 10**(p2[0]*(np.min(eplot)) + p2[1])

tofm = np.max([dcn.calc_TOF(dcn.CBg),TOF_rhs,TOF_lhs])
tofmin = 10**(np.floor((np.min(np.log10(ystatic))-1e-4) / 1) * 1)
tofmax = 10**(np.ceil((np.log10(tofm)+1e-4) / 1) * 1)

elhs = (np.log10(tofmax)-p1[1])/p1[0]
erhs = (np.log10(tofmax)-p2[1])/p2[0]

# Get rate data
r = np.zeros([3,dcn.nfe+1])
for i in range(dcn.nfe+1):
    t = dcn.tdata[i]
    pt = dcn.timeParams(t)
    r[:,i] = np.array([value(dcn.calc_rates(dcn.m,pt,t)[j]) for j in range(3)])

# Rescale rate data to be [-1,1] while preserving signs
rmin = np.min(r,axis=1)
rmax = np.max(r,axis=1)
rscale = np.where(r.T>0,r.T/rmax,np.where(r.T<0,-r.T/rmin,r.T)).T

# Setup everything needed for system plot
posit = {'A': np.array([1/5,4/8]),
         'A*': np.array([2/5,4/8]),
         'B*': np.array([3/5,4/8]),
         'B': np.array([4/5,4/8]),
      }
node_from = ['A','A*','B*']
node_to = ['A*','B*','B']

Callplot = np.vstack([dcn.CAg,dcn.CA,dcn.CB,dcn.CBg])
Csurfplot = np.vstack([dcn.CA,dcn.CB])

old_cmap = plt.cm.get_cmap(mpl.cm.Greens)
new_cmap = old_cmap(np.linspace(0.2,1.0,100))
clipped_cmap = colors.ListedColormap(new_cmap)

cmap1 = mpl.cm.ScalarMappable()
cmap1.set_clim((0,1))
cmap1.set_cmap(clipped_cmap)

cmap2 =  mpl.cm.ScalarMappable()
cmap2.set_clim((-1,1))
cmap2.set_cmap(mpl.cm.coolwarm_r)

fig, ax = plt.subplots()
ax.plot(tplot,Callplot[0,:])
```

```python
# Tile all data to show multiple cycles
ncycles = 1

# Time special case
ttile = np.tile(tplot,(ncycles,1))+np.arange(0,ncycles).reshape(-1,1)
tplot = ttile.reshape(-1)
tticks = np.linspace(tplot[0],tplot[-1],ncycles+1)

# Control Program
Ttile = np.tile(timelist,(ncycles,1))+np.arange(0,ncycles).reshape(-1,1)
Tplot = np.unique(Ttile.reshape(-1))
Etile = np.tile(Elist,(ncycles,1))
Eplot = Etile.reshape(-1)

# Energy Program
etile = np.tile(eplot,(ncycles,1))
eplot = etile.reshape(-1)

# All concs
Calltile = np.tile(Callplot,(1,ncycles))
Callplot = Calltile.reshape(4,-1)

# Surface concs
Csurftile = np.tile(Csurfplot,(1,ncycles))
Csurfplot = Csurftile.reshape(2,-1)

# Scaled Rates rscale
rtile = np.tile(rscale,(1,ncycles))
rscale = rtile.reshape(3,-1)

# Frames
nframes = len(tplot)//5
idframes = np.linspace(0,len(tplot)-1,nframes,dtype=int)

# Network sizing
node_sizes = Callplot*500+500
width_sizes = rscale*2+4
arrow_sizes = np.round(rscale*5+10)

print(Tplot)
print(Eplot)
```

```python
# Verify conc and rates
fig, ax = plt.subplots()
ax.plot(tplot,Callplot[0,:])
```

```python
# Begin generating frames
with tempfile.TemporaryDirectory() as tmpdirname:

    for j in range(nframes):

        index = idframes[j]
        tframe = tplot[index]
        
        # Construct initial figure
        fig = plt.figure(figsize = (8,6)) #W,H
        gs = fig.add_gridspec(6,3, hspace = 1.5, width_ratios = [4,1,5])
        
        ax1 = fig.add_subplot(gs[:2,0])
        ax2 = fig.add_subplot(gs[:3,2])
        ax3 = fig.add_subplot(gs[2:4,0])
        ax4 = fig.add_subplot(gs[4:,0])
        ax5 = fig.add_subplot(gs[3:,1:])
        
        con = mpl.patches.ConnectionPatch(
            (tframe,emax),
            (tframe,-1),
            "data",
            "data",
            axesA = ax1,
            axesB = ax4,
            color = 'r',
            ls = 'dashed',
        )
        fig.add_artist(con)

        # Place energy program plot
        Emod = []
        Tmod = []
        k = 0
        while (tframe > Tplot[k]):
            Emod.append(Eplot[k])
            Tmod.append(Tplot[k])
            k=k+1
        Tmod.append(tframe)
    
        ax1.stairs(Emod,Tmod, baseline=bsln, lw = 2)
        ax1.scatter([tplot[index]],[eplot[index]], c='k', marker='*')
        ax1.plot([tframe,tframe],[emin,emax],'r--')
        ax1.set_xlim([tplot[0],tplot[-1]])
        ax1.set_xticks(tticks)
        ax1.grid()
        ax1.set_ylim([emin,emax])
        #ax1.set_title('Control Program')
        ax1.set_ylabel('BEa (eV)')
        ax1.set_xlabel('Time (s)')
        
        # Draw static TOF plot with energy state marker
        ysint = 10**np.interp(eplot[index],Offsets,np.log10(ystatic))
        
        ax2.semilogy([eplot[index],eplot[index]],[tofmin,tofmax],'r--')
        ax2.semilogy([Offsets[0],elhs],[ystatic[0],tofmax],c='orange',ls='--')
        ax2.semilogy([Offsets[-1],erhs],[ystatic[-1],tofmax],c='orange',ls='--')
        ax2.semilogy(Offsets,ystatic)
        ax2.semilogy([np.min(eplot),np.max(eplot)],[TOF_lhs,TOF_rhs],'k--',marker='*')
        ax2.plot([eplot[index]],[ysint],'k*')
        ax2.set_xlim([ebounds[0]*0.99,ebounds[-1]*1.01])
        ax2.set_ylim([tofmin,tofmax])
        ax2.grid()
        ax2.set_title('Static Performance')
        ax2.set_xlabel('BEa (eV)')
        ax2.set_ylabel('TOF (Hz)')

        # Draw surface species plot
        ax31, = ax3.plot(tplot[:index],Csurfplot[0,:index], label = 'A*')
        ax32, = ax3.plot(tplot[:index],Csurfplot[1,:index], label = 'B*')
        ax3.scatter([tframe*np.ones(2)],Csurfplot[:,index],c='k',marker='*')
        ax3.set_xlim([tplot[0],tplot[-1]])
        ax3.set_ylim([-0.05,1.05])
        ax3.set_xticks(tticks,"")
        ax3.grid()
        #ax3.set_title('Surface Coverage')
        ax3.set_ylabel('Surface Fraction')
        leg = ax3.legend(handles = [ax31,ax32], ncols = 2, loc='center', bbox_to_anchor=(0.5,-0.25))
        #fig.legend(handles = [ax31,ax32] , ncols = 2, loc='center left', bbox_to_anchor=(0.20,0.40))

        # Draw relative rates plot
        ax41, = ax4.plot(tplot[:index],rscale[0,:index],label='A\u2194A*')
        ax42, = ax4.plot(tplot[:index],rscale[1,:index],label='A*\u2194B*')
        ax43, = ax4.plot(tplot[:index],rscale[2,:index],label='B*\u2194B')
        ax4.scatter([tframe*np.ones(3)],rscale[:,index],c='k',marker='*')
        ax4.set_xlim([tplot[0],tplot[-1]])
        ax4.set_ylim([-1.05,1.05])
        ax4.grid()
        ax4.set_xticks(tticks,'')
        #ax4.set_title('')
        #ax4.set_xlabel('Time (s)', alpha = 1)
        ax4.set_ylabel('Scaled Rate (1/s)')
        ax4.legend(handles = [ax41,ax42,ax43], ncol = 3, loc='center', bbox_to_anchor=(0.5,-0.25))
        
        DG = nx.DiGraph()
        
        for i in range(len(node_from)):
            if rscale[i,index] > 0:
                DG.add_edge(node_from[i],node_to[i])
            else:
                DG.add_edge(node_to[i],node_from[i])
        
        dgn = nx.draw_networkx_nodes(DG, posit, ax = ax5, node_size = node_sizes[:,index], node_color = Callplot[:,index], cmap = cmap1.cmap, alpha = 0.8, edgecolors = 'k')
        dgnl = nx.draw_networkx_labels(DG, posit, ax = ax5)
        dge = nx.draw_networkx_edges(DG, posit, ax = ax5, node_size = node_sizes[:,index], edge_color = rscale[:,index], edge_cmap = cmap2.cmap, width = width_sizes[:,index], arrowsize = arrow_sizes[:,index].tolist(), edge_vmin = -1, edge_vmax = 1)
        
        cbar1 = fig.colorbar(cmap1, ax = ax5, pad = 0.01, fraction = 0.10)
        cbar1.set_ticks([])
        cbar1.ax.set_ylabel('Concentration / Surface Coverage')
        
        cbar2 = fig.colorbar(cmap2, ax = ax5, pad = 0.01, fraction = 0.10)
        cbar2.set_ticks([])
        cbar2.ax.set_ylabel('Reaction Rate (1/s)')
        
        #fig.suptitle(f'System at Frame {j:2.0f}/{nframes:3.0f}')
        #ax.set_title(f'Log Reaction Fluxes (1/s)')

        fig.tight_layout()
        fig.savefig(f'{tmpdirname}/RateNetworkTest{j}.png')
        fig.savefig('../ReactionNetworkTesting.png')

        if j == 5:
            fig.savefig('ABVisualLow.eps')
        elif j == 14:
            fig.savefig('ABVisualHigh.eps')
    
        #plt.close()
    
    # Use images in folder to make a GIF
    png_dir = tmpdirname
    images = []
    
    dirFiles = os.listdir(png_dir)
    dirFiles.sort(key=lambda f: int(''.join(filter(str.isdigit, f))))
    
    for file_name in dirFiles:
        if file_name.endswith('.png'):
            file_path = os.path.join(png_dir, file_name)
            images.append(iio.imread(file_path))
    
    #iio.imwrite('TestingFull.gif', images, duration = 5, loop = 0)
    # Duration is ms
```
