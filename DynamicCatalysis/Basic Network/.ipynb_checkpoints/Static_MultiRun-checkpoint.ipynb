```python
import numpy as np
import torch
import random
import matplotlib.pyplot as plt
from pyomo.environ import SolverFactory, value
import matplotlib as mpl

import sys
import os
parent_dir = os.path.abspath(os.path.join(os.path.dirname('ProCatPy'), '..'))
sys.path.insert(0, parent_dir)
from ProCatPy.ProCatSystemFixedFreq import ProCatModel
#from ProCatPy.ProCatOpt import ProCatBO

import warnings
warnings.filterwarnings("ignore")
```

```python
# Establish baseline energy params
seed = 7
np.random.seed(seed)
torch.manual_seed(seed)
random.seed(seed)

eb = np.array([0.7,1.5])
BEParams = np.array([[2],
                     [1.4]])
EaParams = np.array([[0.6],
                     [1]])
no = 51
```

```python
# Generate set of energy params for each simulation
nsim = 10

BEsims = np.random.normal(BEParams,BEParams/25,(2,nsim))
Easims = np.random.normal(EaParams,EaParams/25,(2,nsim))

print(BEsims)
print(Easims)
```

```python
# Set up tracking variables
Offsets = np.zeros([nsim,no])
yA = np.zeros([nsim,no])
yB = np.zeros([nsim,no])
ystaticAg = np.zeros([nsim,no])
ystaticA = np.zeros([nsim,no])
ystaticB = np.zeros([nsim,no])
ystaticBg = np.zeros([nsim,no])
kfs = np.zeros([nsim,no,3])
krs = np.zeros([nsim,no,3])
qs = np.zeros([nsim,no])
r = np.zeros([nsim,no,3])

sX = np.zeros([nsim])
sy = np.zeros([nsim])
```

```python
# Run simulations
for j in range(1):
    
    ns=1
    BEsim = np.array([[2],
                     [1.4]])
    Easim = np.array([[0.6],
                     [1]])
    dcn = ProCatModel(BEsim,Easim)
    #ebounds = np.array([((BEsim[0]-1)*BEsim[1])/BEsim[0]*1.10,BEsim[1]/1.10]).reshape(2)
    ebounds = np.array([[0.7],[2.0]]).reshape(2)
    Offsetlist = np.linspace(ebounds[0],ebounds[1],no).reshape(-1)
    Offsets[j,:] = Offsetlist
    dcn.SetEnergyParams(ebounds,ns,'Square')
    
    # Simulate Model
    for i in range(no):
    
        eparams = {'Amplitude':[0.0],
               'DutyCycle':[0.0],
               'Frequency':1e-2,
               'Offset': Offsetlist[i],
               }
    
        print(f'Sim {j+1:1.0f}| Iter {i+1:3.0f}/{no}| Offset: {Offsetlist[i]:5.3f} eV')
        dcn.SetSignalParams(eparams,False)
        dcn.solveBVP(1,1,'F','Thermo')
        ystaticAg[j,i] = dcn.calc_TOF(dcn.CAg)
        ystaticA[j,i] = dcn.calc_TOF(dcn.CA)
        ystaticB[j,i] = dcn.calc_TOF(dcn.CB)
        ystaticBg[j,i] = dcn.calc_TOF(dcn.CBg)
        yA[j,i] = dcn.CA[0]
        yB[j,i] = dcn.CB[0]
        qs[j,i] = value(dcn.m.q)
    
        pt = dcn.timeParams(0)
        kfs[j,i,:] = pt.kf
        krs[j,i,:] = pt.kr
    
        r[j,i,:] = np.array([value(dcn.calc_rates(dcn.m,pt,0)[j]) for j in range(3)])
        #print(f'{i+1}/{no}:{dcn.calc_TOF(dcn.CBg)}')
    
    sindex = np.argmax(ystaticBg[j,:])
    sX[j] = Offsetlist[sindex]
    sy[j] = ystaticBg[j,sindex]
```

```python
for j in range(1):
    width = 4.8
    fig, ax = plt.subplots(figsize = (width,width*.75))
    ax.plot(Offsets[j,:],yA[j,:])
    ax.plot(Offsets[j,:],yB[j,:])
    ax.legend(['A','B'],fontsize = 10)
    ax.set_xlabel('Binding Energy of A (eV)', weight = 'semibold', size = 10)
    ax.set_ylabel('Surface Fraction', weight = 'semibold', size = 10)
    ax.tick_params(axis='both', which='major', labelsize=10)
    ax.set_xlim(0.8,1.5)
    ax.grid(alpha = 0.6)
    fig.tight_layout()
    fig.savefig('ABVisualStatic.eps')
```

```python

width = 4.8
fig, ax = plt.subplots(figsize = (width,width*.75))
for j in range(1):
    ax.semilogy(Offsets[j,:],ystaticBg[j,:], linewidth = 2, label=f'Sim {j+1}')
    ax.plot([sX[j]],[sy[j]],'r*')

ax.set_xlabel('Binding Energy of A (eV)', weight = 'semibold', size = 10)
ax.set_ylabel('Turnover Frequency (Hz)', weight = 'semibold', size = 10)
ax.tick_params(axis='both', which='major', labelsize=10)

ax.set_xlim([0.7,1.5])
ax.set_ylim([1e-7,None])
ax.grid(alpha = 0.6)
fig.tight_layout()

fig.savefig('SabatierPlot.eps')
```

```python
Ks = kfs/krs

colors = ['blue','orange','green']

for j in range(5):
    fig, (ax1,ax2,ax3) = plt.subplots(1,3,figsize = (14,4))
    for i in range(3):
        ax1.semilogy(Offsets[j,:],kfs[j,:,i],c=colors[i])
        ax1.semilogy(Offsets[j,:],krs[j,:,i],c=colors[i],linestyle = '--')
    ax1.semilogy(Offsets[j,:],np.ones(no),'r--')
    ax1.plot([Es[j,0],Es[j,0]],[0,1e13],'k--')
    ax1.plot([Es[j,1],Es[j,1]],[0,1e13],'k--')
    ax1.set_title('Static Test')
    ax1.set_xlabel('BEa (eV)')
    ax1.set_ylabel('Rate Constant (1/s)')
    #ax.set_xlim(ebounds)
    ax1.legend(['k1f','k1r','k2f','k2r','k3f','k3r'])
    #ax.set_ylim([0,None])
    ax1.grid()

    for i in range(3):
        ax2.semilogy(Offsets[j,:],Ks[j,:,i])
    ax2.semilogy(Offsets[j,:],np.ones(no),'r--')
    ax2.plot([Es[j,0],Es[j,0]],[0,1e10],'k--')
    ax2.plot([Es[j,1],Es[j,1]],[0,1e10],'k--')
    ax2.set_title('Static Test')
    ax2.set_xlabel('BEa (eV)')
    ax2.set_ylabel('Equilibrium Constant')
    #ax.set_xlim(ebounds)
    ax2.legend(['Ag<->A','A<->B','B<->Bg'])
    #ax.set_ylim([0,None])
    ax2.grid()

    for i in range(3):
        ax3.semilogy(Offsets[j,:],r[j,:,i])
    ax3.semilogy(Offsets[j,:],np.ones(no),'r--')
    ax3.plot([Es[j,0],Es[j,0]],[0,1e10],'k--')
    ax3.plot([Es[j,1],Es[j,1]],[0,1e10],'k--')
    ax3.set_title('Static Test')
    ax3.set_xlabel('BEa (eV)')
    ax3.set_ylabel('Reaction Rates')
    ax3.grid()
    #ax.set_xlim(ebounds)
    ax3.legend(['Ag<->A','A<->B','B<->Bg'])    

    # Plot all k ratios and all K ratios
    fig, ax = plt.subplots()
    ax.semilogy(Offsets[j,:],Ks[j,:,0]/Ks[j,:,1])
    ax.semilogy(Offsets[j,:],Ks[j,:,0]/Ks[j,:,2])
    ax.semilogy(Offsets[j,:],Ks[j,:,1]/Ks[j,:,0])
    ax.semilogy(Offsets[j,:],Ks[j,:,1]/Ks[j,:,2])
    ax.semilogy(Offsets[j,:],Ks[j,:,2]/Ks[j,:,0])
    ax.semilogy(Offsets[j,:],Ks[j,:,2]/Ks[j,:,1])
    ax.set_title('K Ratios')
    ax.plot([Es[j,0],Es[j,0]],[0,1e18],'k--')
    ax.plot([Es[j,1],Es[j,1]],[0,1e18],'k--')
    ax.grid()
    ax.legend(['K1/K2','K1/K3','K2/K1','K2/K3','K3/K1','K3/K2'])
    fig.tight_layout()
    
    # # Plot all k ratios
    # for i in range(3):
    #     fig, ax = plt.subplots()
    #     for k in range(3):
    #         ax.semilogy(Offsets[j,:],kfs[j,:,i]/kfs[j,:,k],c=colors[k])
    #         ax.semilogy(Offsets[j,:],kfs[j,:,i]/krs[j,:,k],c=colors[k],linestyle = '--')
    #     ax.plot([Es[j,0],Es[j,0]],[0,1e10],'k--')
    #     ax.plot([Es[j,1],Es[j,1]],[0,1e10],'k--')
    
    #     fig, ax = plt.subplots()
    #     for k in range(3):
    #         ax.semilogy(Offsets[j,:],krs[j,:,i]/kfs[j,:,k],c=colors[k])
    #         ax.semilogy(Offsets[j,:],krs[j,:,i]/krs[j,:,k],c=colors[k],linestyle = '--')
    #     ax.plot([Es[j,0],Es[j,0]],[0,1e10],'k--')
    #     ax.plot([Es[j,1],Es[j,1]],[0,1e10],'k--')

    # Plot forward/reverse of each reaction
    Rg = 0.08314
    T = 423.15
    P0 = 1.01325

    r1f = kfs[j,:,0]*0.99*Rg*T/P0*(1-yA[j,:]-yB[j,:])
    r1r = krs[j,:,0]*yA[j,:]
    r2f = kfs[j,:,1]*yA[j,:]
    r2r = krs[j,:,1]*yB[j,:]
    r3f = kfs[j,:,2]*yB[j,:]
    r3r = krs[j,:,2]*0.01*Rg*T/P0*(1-yA[j,:]-yB[j,:])

    fig, ax = plt.subplots()
    ax.semilogy(Offsets[j,:],r1f)
    ax.semilogy(Offsets[j,:],r1r)
    ax.semilogy(Offsets[j,:],r2f)
    ax.semilogy(Offsets[j,:],r2r)
    ax.semilogy(Offsets[j,:],r3f)
    ax.semilogy(Offsets[j,:],r3r)
    ax.plot([Es[j,0],Es[j,0]],[0,1e3],'k--')
    ax.plot([Es[j,1],Es[j,1]],[0,1e3],'k--')
    ax.grid()
    ax.legend(['r1f','r1r','r2f','r2r','r3f','r3r'])
```
